# ANALYSIS

```python
# IDENTIFY hIGHEST MONTHLY SALES AND LOWEST MONTHLY SALES 

Monthly_sales_Analysis = ecom.groupby('order_month')['Sales'].sum().reset_index ()# to order use .sort_values(ascending=False) for highest _ascending true for lowest 
```

```python
Monthly_sales_Analysis.sort_values(by='Sales',ascending=True)
```

```python
# visualizing 
fig= px.line(Monthly_sales_Analysis,
             x='order_month',
             y='Sales',
             title= 'Monthly_sales_Analysis')
fig.show()
```

## --November has the highest sales 
## --February  has the lowest sales

## Analyse sales and profit by customer segment 

```python
sales_profit_by_segment=ecom.groupby('Segment').agg({'Sales':'sum','Profit':'sum'}).reset_index()
fig = go.Figure()

color_palette=color.qualitative.Pastel

fig.add_trace(go.Bar(x=sales_profit_by_segment['Segment'], 
                     y=sales_profit_by_segment['Sales'], 
                     name='Sales',
                     marker_color=color_palette[0]))

fig.add_trace(go.Bar(x=sales_profit_by_segment['Segment'], 
                     y=sales_profit_by_segment['Profit'], 
                     name='Profit',
                     marker_color=color_palette[1]))

fig.update_layout(title='Sales and Profit Analysis by Customer Segment',
                  xaxis_title='Customer Segment', yaxis_title='Amount')


fig.show()
```

## --Sales to profit ratio on basis of customer segment 

```python
 sales_profitratio =ecom.groupby('Segment').agg({'Sales':'sum','Profit':'sum'}).reset_index()
 sales_profitratio['sales-profit_ratio']=sales_profitratio['Sales']/sales_profitratio['Profit']
sales_profitratio
```

```python
fig=px.bar(sales_profitratio,x='Segment',y='sales-profit_ratio',title='Sales-to-Profit_ratio')
fig.show()
```

## Profit margin ratio on customer segment 

```python
sales_profit= ecom.groupby('Segment').agg({'Sales':'sum','Profit':'sum'}).reset_index()
sales_profit['Profitmargin-Ratio']=sales_profit['Profit']/sales_profit['Sales']
print(sales_profit)
```
